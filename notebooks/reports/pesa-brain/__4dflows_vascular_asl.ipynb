```python
import os
import sys
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path(os.path.abspath('')).resolve().parents[2]
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))


import nvitk as nv
from nvitk import db
```

```python
tmp_vascular_att = pd.read_csv('/data_local/LabVF/PESA-Brain/DB/new_vars/Vascular_Atlas_ATT_Results.csv')
tmp_vascular_att
```

```python
db_vascular_att = tmp_vascular_att.pivot_table(
    index='mri_id',
    columns=['territory', 'smoothing'],
    values='mean_att',
    aggfunc='mean'
)

# Flatten the MultiIndex columns to match your 'roi-s' format
db_vascular_att.columns = [f"{col[0]}-{col[1]}" for col in db_vascular_att.columns]

# Add patient_id from db_asl_cerebral_perfusion that matches mri_id
db_vascular_att['mri_id'] = pd.Series(pd.NA)
mriid2pesaid = {row['mri_id']: row.name for _, row in db_vascular_att.iterrows()}
db_vascular_att['patient_id'] = db_vascular_att.index.map(mriid2pesaid)
db_vascular_att
```

```python
db_vascular_att = tmp_vascular_att.pivot_table(
    index='mri_id',
    columns=['territory', 'smoothing'],
    values='median_att',
    aggfunc='median'
)

# Flatten the MultiIndex columns to match your 'roi-s' format
db_vascular_att.columns = [f"{col[0]}-{col[1]}" for col in db_vascular_att.columns]

# Add patient_id from db_asl_cerebral_perfusion that matches mri_id
db_vascular_att['mri_id'] = pd.Series(pd.NA)
mriid2pesaid = {row['mri_id']: row.name for _, row in db_vascular_att.iterrows()}
db_vascular_att['patient_id'] = db_vascular_att.index.map(mriid2pesaid)
db_vascular_att
```
