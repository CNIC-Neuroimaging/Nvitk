```python
from __future__ import annotations

import pandas as pd
from nvitk.db import DataRepo
from nvitk.db.storage import utc_now_iso

# --- config ---
PIPELINE_ID = "4dflow_v2"  
DERIVED_VAR = "psf"
SOURCE_FILE = "derived"
SOURCE_SHEET = "psf"
SOURCE_COLUMN = "psf"  # stable upsert key together with source_file/sheet


def derive_psf_from_flow_tseries(repo: DataRepo, *, pipeline_id: str = PIPELINE_ID) -> pd.DataFrame:
    """Max(value_num) over frames per (subject, session, region) for variable flow_tseries."""
    df = repo.get(
        "image_measurements",
        filters={
            "modality": "4dflow",
            "variable_id": "flow_tseries",
            "pipeline_id": pipeline_id,
        },
    )
    if df.empty:
        return df

    gcols = ["subject_uid", "session_id", "region_id", "region_label", "pipeline_id"]
    gcols = [c for c in gcols if c in df.columns]
    sub = df.copy()
    sub["value_num"] = pd.to_numeric(sub["value_num"], errors="coerce")
    sub = sub.loc[sub["value_num"].notna()]

    agg = (
        sub.groupby(gcols, dropna=False, as_index=False)["value_num"]
        .max()
        .rename(columns={"value_num": "value_num"})
    )
    return agg


def build_psf_measurement_rows(agg: pd.DataFrame, *, pipeline_id: str, source_batch_id: str = None) -> pd.DataFrame:
    """Rows matching image_measurements shape for upsert."""
    n = len(agg)
    out = pd.DataFrame(
        {
            "subject_uid": agg["subject_uid"].astype("string"),
            "session_id": agg["session_id"].astype("string") if "session_id" in agg.columns else pd.Series([""] * n, dtype="string"),
            "modality": pd.Series(["4dflow"] * n, dtype="string"),
            "region_id": agg["region_id"].astype("string") if "region_id" in agg.columns else pd.Series([pd.NA] * n, dtype="string"),
            "region_label": agg["region_label"].astype("string") if "region_label" in agg.columns else pd.Series([pd.NA] * n, dtype="string"),
            "frame_index": pd.Series([pd.NA] * n, dtype="Int64"),
            "variable_id": pd.Series([DERIVED_VAR] * n, dtype="string"),
            "value_num": agg["value_num"].astype("float64"),
            "value_text": pd.Series([pd.NA] * n, dtype="string"),
            "unit": pd.Series([pd.NA] * n, dtype="string"),
            "value_kind": pd.Series(["float"] * n, dtype="string"),
            "pipeline_name": pd.Series([f"derived_{DERIVED_VAR}"] * n, dtype="string"),
            "pipeline_id": pd.Series([pipeline_id] * n, dtype="string"),
            "qc_status": pd.Series([pd.NA] * n, dtype="string"),
            "source_asset": pd.Series([pd.NA] * n, dtype="string"),
            "source_table": pd.Series([f"{SOURCE_FILE}::{SOURCE_SHEET}"] * n, dtype="string"),
            "source_file": pd.Series([SOURCE_FILE] * n, dtype="string"),
            "source_sheet": pd.Series([SOURCE_SHEET] * n, dtype="string"),
            "source_column": pd.Series([SOURCE_COLUMN] * n, dtype="string"),
            "source_batch_id": pd.Series([source_batch_id] * n, dtype="string"),
            "measured_at": pd.Series([pd.NaT] * n, dtype="datetime64[ns]"),
            "updated_at": pd.Series([utc_now_iso()] * n, dtype="string"),
        }
    )
    return out


def register_psf_variable(repo: DataRepo) -> None:
    repo.register_variables(
        [
            {
                "variable_id": DERIVED_VAR,
                "source_column": SOURCE_COLUMN,
                "source_file": SOURCE_FILE,
                "source_sheet": SOURCE_SHEET,
                "aliases": [DERIVED_VAR, "PSF", "peak_systolic_flow"],
                "domain": "image",
                "table": "image_measurements",
                "modality": "4dflow",
                "label": "Peak systolic flow (max over flow_tseries frames)",
                "value_kind": "float",
            }
        ]
    )


# --- run ---
repo = DataRepo()  # or DataRepo("/path/to/dataset")

agg = derive_psf_from_flow_tseries(repo, pipeline_id=PIPELINE_ID)
if agg.empty:
    raise RuntimeError("No flow_tseries rows; check pipeline_id and import.")

psf_df = build_psf_measurement_rows(agg, pipeline_id=PIPELINE_ID)
psf_df
```
